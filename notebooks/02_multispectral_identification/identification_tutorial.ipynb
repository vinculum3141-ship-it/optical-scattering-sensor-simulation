# 02 — Multi-Spectral Material Identification

## Objective

Identify a material from its spectral reflectance measured at several wavelengths: build a spectral stack, extract per-pixel spectral vectors, and classify each pixel against a small reference library.

## What you'll see

- A synthetic multispectral image with two distinct material regions (coating vs. substrate)
- `SpectralAnalyzer` classification — spectral angle mapping and band ratios — with a dominant label per pixel

## How to use this notebook

Change the parameters in the next cell and rerun the later cells to see how the classification changes.

```python
import numpy as np

from optical_metrology.analysis import ReferenceSpectrum, SpectralAnalyzer
from optical_metrology.detector import DigitalImage
```

```python
wavelengths = np.array([450.0, 550.0, 650.0])
reference = np.array([0.85, 0.15, 0.05])
background = np.array([0.10, 0.20, 0.80])

pixels = np.zeros((32, 32, 3), dtype=float)
pixels[:, :, :] = background
pixels[8:24, 8:16, :] = reference
pixels[8:24, 16:24, :] = np.array([0.25, 0.45, 0.25])

image = DigitalImage(pixels=pixels, metadata={"bit_depth": 12, "wavelengths": wavelengths})
print(image.visualize())
```

```python
analyzer = SpectralAnalyzer(
    reference_library=[
        ReferenceSpectrum("coating", reference, wavelengths),
        ReferenceSpectrum("substrate", background, wavelengths),
    ]
)
report = analyzer.analyze(image)
print(report.measurements)
```
